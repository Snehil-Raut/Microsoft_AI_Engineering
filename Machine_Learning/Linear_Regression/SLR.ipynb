```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
# Instead of calling Linear Regression from sklearn - build the own algo

# Getting the values of m & b
```

```python
class ownLR:
  def __init__(self) -> None:
    self.m = None
    self.b = None

  def train(self,X_train, y_train):
    # calculate value of m and b
    num=0
    den=0
    for i in range(len(X_train)):
      # this repeats 160 time
      num += (X_train[i] - X_train.mean()) * (y_train[i] - y_train.mean())
      den += (X_train[i] - X_train.mean()) * (X_train[i] - X_train.mean())

    self.m = num/den
    self.b = y_train.mean() - self.m * X_train.mean()

    print("Slope(m)", self.m)
    print("Intercept (b)", self.b)

  def predict(self,X_test):
    return self.m * X_test + self.b
```

```python
LR = ownLR()

# object of Linear regression
```

```python
LR.train(X_train, y_train)
```

```python
# now this is train we can predict
```

```python
X_test[0]  # first cgpa is 8.568
```

```python
LR.predict(X_test[0])  # model predicted 3.89 salary
```
