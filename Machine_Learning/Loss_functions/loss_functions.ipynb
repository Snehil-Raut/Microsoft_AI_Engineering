# Mean Squared Error
- R² = 1 − (SS_res / SS_tot)
- SS_res = residual sum of squares    SUM(y_test - y_pred)^2
- SS_tot = total error of guess       SUM(y_test - mean(y_test))^2

```python
# MEAN SQUARED ERROR in Python

# formula = 1- ss(residulas) / ss(total)

y_test = [10,20,30,40,50]  # actual values
y_pred = [11,19,31,39,51]  # model's guess

# Get the residual error (y_test - y_pred)
residual_error = []

for error in zip(y_test, y_pred):
    residual_error.append(sum(error))

mean_y_test = sum(y_test) / len(y_test)
print("Mean of y_test is",mean_y_test)    # Model should beat this score

# Now ss_total-  SUM(y_test-mean)^2
ss_total = []
for ytest in y_test:
    gap = ytest - mean_y_test     # subtract y_test mean with each y_test(actual value)
    ss_total.append(gap**2)       # append the square of gap between them
print("Total Error from ss_tot is", sum(ss_total))

# Now ss_residuals - SUM(y_test - y_pred)^2
ss_res = []
for ytest, ypred in zip(y_test, y_pred):
    gap = ytest - ypred
    ss_res.append(gap**2)
print("Total Error from ss_res is", sum(ss_res))

# R^2 score

# r^2 = 1 - (ss_res / ss_tot)
r2 = 1 - (sum(ss_res) / sum(ss_total))
print("R^2 Score is", r2)
print(f"Model explains: {r2*100}% variation in score")
```

```python
# Verify the above output with scikit learn library
from sklearn.metrics import r2_score

y_test = [10,20,30,40,50]  # actual values
y_pred = [11,19,31,39,51]  # model's guess

r2 = r2_score(y_test, y_pred)

print(f"r2 score is {r2}")
print(f"Model explains: {r2*100}% variation in score")
```

# Root mean squared error

- Formula:

- RMSE = √( Σ(yᵢ − ŷᵢ)² / n )
- yᵢ = actual (y_test)
- ŷᵢ = predictated(y_pred)
- n = total values in y_test

```python
# PYTHON CODE

import math
y_test = [10,20,30,40,50]  # actual values
y_pred = [11,19,31,39,51]  # model's guess

errors = []
for ytest, ypred in zip(y_test, y_pred):
    error = ytest - ypred
    errors.append(error**2)

# Get the mean of y_test, y_pred 
mean_error = sum(errors) / len(y_test)

print("Mean error is", mean_error)

# RMSE - root of MSE
rmse = math.sqrt(mean_error)
print("RMSE is", rmse)
```

```python
# Verify with sklearn

from sklearn.metrics import root_mean_squared_error
y_test = [10,20,30,40,50]
y_pred = [11,19,31,39,51]


rmse = root_mean_squared_error(y_test, y_pred)
print("RMSE is", rmse)
```

# Mean Absolute error

- MAE = Σ|yᵢ − ŷᵢ| / n
- yᵢ = actual (y_test)
- ŷᵢ = predicted(y_pred)
- n = total values in y_test

```python
# PYTHON CODE

# get mean of y_test - y_pred
y_test = [10,20,30,40,50]
y_pred = [11,19,31,39,51]

abs_errors = []
for ytest, ypred in zip(y_test, y_pred):
    error = ytest - ypred
    abs_errors.append(abs(error))       # append the ABSOLUTE VALUE OF THE ERROR EVERYTIME

# Sum of |abs_error| / n
mae = sum(abs_errors) / len(y_test)

print("Mean absolute error is", mae)
```

```python
# Verify with sklearn
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(y_test, y_pred)
print("Mean absolute error is", mae)
```
