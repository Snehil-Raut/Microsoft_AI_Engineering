# Implement simple gradient descent to minimize a quadratic function. Plot how loss decreases each step.

```python
#  y= mx + c

import matplotlib.pyplot as plt

def loss_function(x):
    return (x-3)**2    # f(x) = (x-3)^2

def gradient(x):  
    return 2*(x-3)     # Defines the slope of hill at any x value

x = 10
lr = 0.1   # learning rate
losses = []
gradient_x = []
n_steps = 50   # more steps higher chances to reach minimum

for i in range(n_steps):
    
    loss = loss_function(x)            # get the loss at each step
    print("Loss",loss)
    gradient_descent = gradient(x)     # calculate the gradient
    print("gradient",gradient_descent)

    x = x - lr * gradient_descent       # find new x
    print("new x",x)

    losses.append(loss)
    gradient_x.append(x)

# Loss drop
fig,ax = plt.subplots(1,2, figsize=(12,5))
ax[0].plot(losses)
ax[0].set_title('Losses per step')
ax[0].set_xlabel('Step')
ax[0].set_ylabel('Loss')

# U curve
x_curve = [i * 0.1 for i in range(-80, 150)]    # x values for the curve
y_curve = [loss_function(x) for x in x_curve]   # for each x in x_curve
ax[1].plot(x_curve, y_curve)
ax[1].scatter(gradient_x, [loss_function(x) for x in gradient_x], color='red')
ax[1].set_title('Ball rolling down the hill')
ax[1].set_xlabel('x')
ax[1].set_ylabel('f(x)')

plt.tight_layout()
plt.show()
```
