```python
import pandas as pd
import numpy as np
```

# 1. Load & Explore

```python
df = pd.read_json('e-comm_data.json')
df
```

```python
# check dtype, shape and missing/null value
print("Shape",df.shape)
print()
print("Datatype",df.dtypes)
print()
print("Null values", df.isna().sum())
```

```python
df.head()
```

# 2 Dealing with duplicates

```python
# exact duplciates
duplicate_rows = df[df.duplicated(keep=False)]
duplicate_rows
```

```python
# Drop duplicate and keep 1st occurance of each

df.drop_duplicates(keep='first', inplace=True)
df
```

```python
# how to work with hannah lee
```

# 3. Customer Name Cleaning

```python
# Fix inconsistent name casing, extra white space from all object columns

obj_col = df.select_dtypes(include=['object']).columns

# apply fix on each object col
df[obj_col] = df[obj_col].apply(lambda x: x.str.strip().str.title())

# fill Null customer with Unknown
df['customer_name'] = df['customer_name'].fillna('Unknown')
df
```

```python
# Make dates to YYYY-MM-DD

# making dates in 1 format
df['order_date'] = df['order_date'].str.replace(r'[\.,/]', '-', regex=True)

print("Datatype before:",df['order_date'].dtype)

df['order_date']
```

```python
# get the data and convert to date-time format and follow YYYY-MM-DD format
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed', errors='coerce')
print("Data type after:",df['order_date'].dtype)
df['order_date']

#df['order_date'] = df['order_date'].dt.strftime('%Y-%m-%d') - if want to convert date col to object
```

```python
# make none/null to pending in delivery_date

df['delivery_date'] = pd.to_datetime(df['delivery_date'],  errors='coerce')
df['delivery_date']
```

```python
df.dtypes
```

# 5. Logical Date Validation

```python
# Find orders where delivery_date is before order_date 
df['is_date_valid'] = np.where(df['delivery_date'] < df['order_date'], "Invalid date", "Valid date")

# return rows which has Invalid date
invalid_dates = df.loc[df['is_date_valid'] == 'Invalid date']
invalid_dates
```

# 6. Quantity Validation

```python
# Negative quantity — (ORD012)
# Zero quantity — can an order exist with 0 items? (ORD002)


# replace Negative and zero value with Nan
df.loc[df['quantity'] <= 0 , 'quantity'] = np.nan

# convert the column back to int and replace nan with NA
df['quantity'] =df['quantity'].astype('Int64')

df['quantity']
```

```python
# Flag Suspiciously high quantity like 999 — flag as outlier (ORD007)

suspicious_quantities = df[df['quantity'] > 100]
suspicious_quantities
```

# 7 Discount validating

```python
# Discount must be between 0 and 100 percent
# Find discounts over 100 or negative discounts
# Replace invalid discounts with NaN

df['discount_pct']
```

```python
df.loc[(df['discount_pct'] > 100) | (df['discount_pct'] < 0), 'discount_pct' ] = np.nan

# convert column back to int
df['discount_pct'] =df['discount_pct'].astype('Int64')

df['discount_pct']
```

# 8 Price Validation

```python
# Find negative unit_price — replace with NaN
df['unit_price']
```

```python
df.loc[df['unit_price'] < 0, 'unit_price'] = np.nan

df['unit_price'] =df['unit_price'].astype('Int64')

df['unit_price']
```

# 9. Rating Validation

```python
df['rating']
```

```python
# <!-- Valid range is 1.0 to 5.0
# Find ratings of 0.0, 6.0, 99.0 — replace with NaN -->

lower_rating = 1.0
highest_rating = 5.0

# using masking
df['rating'] = df['rating'].mask((df['rating'] < lower_rating) | (df['rating'] > highest_rating))
df['rating']
```

 # 10 New column for total_price

```python
# New total price column
df['total_price'] = df['quantity'] * df['unit_price'] * (1 - df['discount_pct'] / 100)
df['total_price']

# pandas handels NaN value automatically - if any of the value is NaN, the output is NaN
```
