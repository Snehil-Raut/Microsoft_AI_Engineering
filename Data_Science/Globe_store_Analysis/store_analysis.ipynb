```python
# ## Q31 Build a simple RFM (Recency, Frequency, Monetary) summary per customer using Order Date and Sales

# gives info on Day after the last purchase date
reference_date = df['Order Date'].max() + pd.Timedelta(days=1)
rfm_df = df.groupby('Customer ID').agg({
    'Order Date' : lambda x:(reference_date - x.max()).days,

    # Frequency of orders
    'Order ID' : 'nunique',

    # Momentart - total spending across all transactions
    'Sales' : 'sum'
}).reset_index()

rfm_df.columns = ['Customer ID', 'Recency', 'Frequency', 'Monetary']

rfm_df.head(10)

# reference date = 1 day after the last order 
# Recency - how many days ago The customer last ordered
# Frequence - how many distinct orders they place
# Monetary - total money spent
```

# Phase 6 — Advanced & business insights

```python
### Q38 What % of total profit comes from the top 20% of products? (80-20 analysis)

unique_products = df.groupby('Product Name')['Profit'].sum().reset_index()

# sort products Sales wise
unique_products = unique_products.sort_values(by='Profit', ascending=False)

# Calculate total profit and get 20% of products
total_profit = unique_products['Profit'].sum()
top_20_products = int(len(unique_products) * 0.20)  # gets the rows of top 20% products

# Add up to profit of 20% and compute %
top_20_profit = unique_products.head(top_20_products)['Profit'].sum()
percentage_profit = (top_20_profit / total_profit) * 100

print(f"Percentage of top 20% Products Profit is {percentage_profit}%")
```

```python
### Q39 Calculate YoY sales growth for each year. Is the business growing?

Sales_growth = df.groupby(df['Order Date'].dt.year)['Sales'].sum().round(3).reset_index()

plt.figure(figsize=(8,4))
plt.plot(Sales_growth['Order Date'], Sales_growth['Sales']/100000, linestyle='-', marker='o', linewidth=2)
plt.title('YOY Sales growth')
plt.xlabel('Years')
plt.ylabel('Sales in Lacs')
plt.show()
```

```python
### Q40 Which Sub-Category has high profit margin AND high volume(orders)? Find the sweet spot.

sweet_spot = df.groupby('Sub-Category').agg({
    'Profit Margin' : 'mean',
    'Order ID': 'nunique'
}).sort_values(by=['Profit Margin', 'Order ID'], ascending=False).reset_index()

sweet_spot
```

```python
### Q41 Group orders into discount buckets. Compare average profit margin per bucket.


discount_order_buckets = pd.cut(df['Discount'], bins=[0, 0.01, 0.20, 0.50, 1.0], include_lowest=True) # include_lowest=True, ensures the lowest value fall into 1st category
df.groupby(discount_order_buckets)['Profit Margin'].mean()

# High discount leads to low profit margin
```

```python
### Q42 Is there a relationship between Quantity ordered and Discount given? Do bulk orders get bigger discounts?
quantity_vs_discount = df.groupby('Quantity')['Discount'].mean().reset_index()
mean_y = np.mean(quantity_vs_discount['Discount'])
plt.figure(figsize=(20,4))

plt.scatter(quantity_vs_discount['Quantity'], quantity_vs_discount['Discount'], color='steelblue', alpha=0.7 )
plt.axhline(y=mean_y,color='red', linestyle='--', label= f'Discount Mean ({mean_y:.3f})' )
plt.title('Quantity & Discount Relation')
plt.xlabel('Quantity')
plt.ylabel('Discount')
plt.legend()
plt.show()

# there's no relation ,as some orders with low quantity has higher discount and vice versa
```
