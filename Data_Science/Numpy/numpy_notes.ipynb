**Matrix Properties**
- Rank(Mehtod 1) - No. of non-zero rows
- Determinant (Method 2) - If det is not 0 , Rank is 3, If det is 0 , Rank<3

* If Det==0, means Rank<3, What is the exact Rank? (3x3 matrix)
- Take any 2x2 elements and find Det. If Det=0, then Rank=1. If Det not equal to 0, Rank=2
